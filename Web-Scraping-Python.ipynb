# Web Scraping – U.S. Companies by Revenue

A Python web scraping project using Requests, BeautifulSoup, and Pandas to extract company data from Wikipedia.


```python
import numpy as np
import pandas as pd
import requests
from bs4 import BeautifulSoup
```

# 1- Request

```python
url = "https://en.wikipedia.org/wiki/List_of_largest_companies_in_the_United_States_by_revenue"
```

```python
headers = {"User-Agent": "Learning Web Scraping with python"}
```

```python
page = requests.get(url , headers=headers)
```

```python
page
```

# 2- Parse

```python
soub = BeautifulSoup(page.text)
```

```python
soub
```

```python
soub.find_all('table')
```

```python
table = soub.find_all('table')[2]
```

```python
table
```

# 3- Extracting

```python
colheader = table.find_all('th')
colheader
```

```python
colheader = [col.text.strip() for col in colheader]
df = pd.DataFrame(columns= colheader)
```

```python
df
```

```python
coldata = table.find_all('tr')
```

```python
coldata 
```

```python
for row in coldata [1:] :
    rowdata = row.find_all('td')
    cells = [data.text.strip()for data in rowdata]
    length = len(df)
    df.loc[length]= cells
```

```python
df
```

```python
df.to_csv("List-of-Companies-by-Revenue.csv", index=False)
```
